```python
import numpy as np
```

```python
input_samples = np.array([[ 1, -1,  1],
                          [-1,  1, -1],
                          [ 1,  1,  -1]])
```

```python
weights = np.zeros( (input_samples[0].shape[0],input_samples[0].shape[0]) )
weights
```

```python
def apply_heb(row, weights):
    num = row.shape[0]
    row = row.reshape( (1, num) )
    rowT = row.reshape( (num, 1) )
    
    mat = rowT @ row
    np.fill_diagonal(mat, 0)
    weights += mat
```

```python
for row in input_samples:
    apply_heb(row, weights)

weights
```
