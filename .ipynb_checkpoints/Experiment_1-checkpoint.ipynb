## Objective

To design a perceptron model that computes the weighted sum of input features and applies a threshold-based activation function to generate a binary output.

---

## Theory

A Perceptron is a fundamental building block of Artificial Neural Networks. It performs a linear combination of input features followed by a step (threshold) activation function.

Mathematically, the perceptron computes:

z = w·x + b

where:
- w = weight vector  
- x = input feature vector  
- b = bias  

The output is determined using a step activation function:

Output = 1,  if z ≥ threshold  
Output = 0,  otherwise

---

## Input Used

- Number of input features: 2
- Input values are randomly generated using NumPy
- Each execution produces a different input sample

---

## Algorithm

1. Accept a single input vector
2. Initialize weights and bias
3. Compute linear sum: z = w·x + b
4. Apply step activation using threshold
5. Return binary output (0 or 1)

---

## Code:

```python
# implement perceptron function

class Perceptron:
    def __init__(self, activation_threshold = 0):
        # set the thresehold limit of the step activation function
        self.threshold = activation_threshold

    def forward(self, input_x, verbose = False):

        # check if input shape is correct (1 dimensional)  i.e. 1 instancea at a time
        if len(input_x.shape) != 1:
            raise Exception("invalid input point")

        # initialise parameters
        w = np.ones(input_x.shape)
        b = np.zeros((1,))
        if verbose:
            print(f"Weight: {w}\nBias: {b}")

        # forward calculation
        z = np.dot(input_x,w) + b
        if verbose:
            print(f"Linear_Sum: {z}")
            
        # step activation function
        return 1 if z >= self.threshold else 0
        
```

```python
import numpy as np
# random number generetor to use randomness in training
rng = np.random.default_rng()

# 2 input features here
input_x = rng.standard_normal((2,))
print(f"Input: {input_x}")

# model instance
model = Perceptron()

# output
output_y = model.forward(input_x, verbose = True)

print(f"Output: {output_y}")
```


## Observations

- The perceptron computes a linear combination of input features
- Output depends on whether the linear sum crosses the threshold
- Since weights are fixed (all ones), the decision boundary is a straight line
- No learning occurs because weights are not updated
