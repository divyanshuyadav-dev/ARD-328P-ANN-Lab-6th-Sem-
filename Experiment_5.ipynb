## Objective

To write a Python program that implements different activation functions and plots their graphs for visualization and comparison.

---

## Theory

Activation functions introduce non-linearity into neural networks. Without activation functions, a neural network would behave like a linear model regardless of the number of layers.

In this experiment, the following activation functions are implemented:

### 1. Step Function

f(x) = 1,  if x > threshold  
f(x) = 0,  otherwise

- Produces binary output  
- Used in perceptrons  
- Not differentiable  

---

### 2. Sigmoid Function

σ(x) = 1 / (1 + e^(−x))

- Output range: (0, 1)  
- Smooth and differentiable  
- Commonly used in binary classification  
- Suffers from vanishing gradient problem  

---

### 3. Hyperbolic Tangent (tanh)

tanh(x) = (e^x − e^(−x)) / (e^x + e^(−x))

- Output range: (−1, 1)  
- Zero-centered  
- Steeper than sigmoid  
- Can suffer from vanishing gradient  

---

### 4. ReLU (Rectified Linear Unit)

f(x) = max(0, x)

- Output range: [0, ∞)  
- Computationally efficient  
- Does not saturate for positive inputs  
- Can suffer from dying ReLU problem  

---

## Input Range

- Input values generated from −10 to 10  
- Step size: 0.1  

---

## Code:

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# define activation functions

def step(x, threshold = 0):
    return 1 if x > threshold else 0

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    pos_exp = np.exp(x)
    neg_exp = np.exp(-x)
    return (pos_exp - neg_exp) / (pos_exp + neg_exp)

def relu(x):
    return x if x > 0 else 0

# activation functions as list
activation_functions = [step, sigmoid, tanh, relu]
```

```python
# fucntion to plot graph
def make_graph(x, functions, axes):
    # assume no of axes equals-to no of functions, i.e. one axes for each activation function
    for idx in range(len(axes)):
        # for each input x apply activation function according to its subplot axes
        axes[idx].plot(x, np.array( [ functions[idx](x_idx)  for x_idx in x ] ) , 'r-')
        # set subplot title accrd to activation func.
        axes[idx].set_title(functions[idx].__name__)
        axes[idx].set_xlabel("Input: X")
        axes[idx].set_ylabel(f"Ouput: {functions[idx].__name__}(X)")
        axes[idx].grid(True)
```

```python
# define inputs
input_x = np.arange(-10,10,.1)

# extract subplots
fig, axes = plt.subplots(2,2, figsize=(17.5, 12.5))
# reshape axes accord. to function requirements
flat_axes = axes.reshape(1,4)

# # plot graph
make_graph(input_x, activation_functions, flat_axes[0])
```

## Observations

- The step function shows a sudden jump at the threshold.
- Sigmoid produces an S-shaped curve bounded between 0 and 1.
- Tanh produces an S-shaped curve centered at zero.
- ReLU outputs zero for negative inputs and increases linearly for positive inputs.
